## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

### Baseline rule

I will prioritize content for refresh when it shows a combination of staleness, declining trend, and search opportunity.

The rule uses:
- days_since_last_update as the staleness signal
- trend_direction as the performance-direction signal
- search_volume as the opportunity signal

A higher score means higher refresh priority. This is a decision-support baseline, not a claim that every high-scoring page requires a refresh.

```python
# Signal check 1: staleness

df["staleness_bucket"] = pd.cut(
    df["days_since_last_update"],
    bins=[-1, 90, 180, 365, float("inf")],
    labels=["0-90", "91-180", "181-365", "365+"]
)

staleness_check = (
    df.groupby("staleness_bucket", observed=False)
      .agg(
          n=("content_id", "size"),
          avg_impressions_90d=("impressions_90d", "mean"),
          avg_ctr=("ctr", "mean"),
          avg_trend_pct=("trend_pct", "mean")
      )
      .reset_index()
)

display(staleness_check)
```

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
import numpy as np

# Normalize the two numeric opportunity signals to a 0-100 scale.
staleness_score = (
    df["days_since_last_update"]
    / df["days_since_last_update"].max()
) * 100

search_score = (
    df["search_volume"].fillna(0)
    / df["search_volume"].max()
) * 100

# Give full trend points only to pages with a declining trend.
trend_score = np.where(
    df["trend_direction"].eq("down"),
    100,
    0
)

# One simple additive baseline score.
df["baseline_score"] = (
    0.40 * staleness_score
    + 0.30 * search_score
    + 0.30 * trend_score
)

# One reason code and one action label.
df["reason_code"] = "STALE_DECLINING_CONTENT"
df["action_label"] = "REFRESH_CONTENT"

# Rank highest score first.
queue = (
    df.sort_values(
        ["baseline_score", "content_id"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

queue["rank"] = queue.index + 1

display(
    queue[
        [
            "rank",
            "content_id",
            "baseline_score",
            "reason_code",
            "action_label",
            "days_since_last_update",
            "search_volume",
            "trend_direction"
        ]
    ].head(20)
)
```

```python
import os

output_dir = "/content/flyrank-ml-internship-haseeb/work/outputs"
os.makedirs(output_dir, exist_ok=True)

output_path = os.path.join(
    output_dir,
    "baseline_action_score.csv"
)

queue.to_csv(output_path, index=False)

print(f"Saved: {output_path}")
print(f"Rows written: {len(queue):,}")
```

### Weak picks

The weakest-ranked pages are lower priority under this baseline because they have a weaker combination of staleness, search opportunity, and declining trend.

These are not necessarily pages that should never be refreshed. A low score only means that the simple baseline does not prioritize them.

The main weakness observed in the top-ranked results is that some pages receive high scores because they are very stale and trending down even when search volume is zero or missing. This means the baseline can over-prioritize stale pages with limited or uncertain search opportunity.
